```python
import os, sys
if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")
sys.path.insert(0, os.getcwd())

import pandas as pd
from src.classifier.data import CATEGORIES
from src.classifier.split import make_split, save_split, load_split, apply_split

df = pd.read_csv("data/opp115_segments.csv")
```

```python
split = make_split(df, seed=42)
save_split(split)
print("seed used:", split["seed"])
```

```python
split = load_split()
parts = apply_split(df, split)

# no policy in two sets
a, b, c = (set(split[k]) for k in ("train", "val", "test"))
assert not (a & b) and not (a & c) and not (b & c)

summary = pd.DataFrame({k: v[CATEGORIES].sum() for k, v in parts.items()})
summary.loc["segments"] = [len(v) for v in parts.values()]
summary.loc["policies"] = [len(split[k]) for k in ("train", "val", "test")]
summary
```
